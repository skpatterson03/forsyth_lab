```python
import gffpandas.gffpandas as gffpd
import pandas as pd
import numpy as np
```

```python
#all dataframes from previous working steps
merged = pd.read_csv("methylation_df.csv")
ctrl_unique = pd.read_csv('./methyl_point_diff/ctrl_diff_all.csv')
ars_unique = pd.read_csv('./methyl_point_diff/arsS_diff_all.csv')
ph5_unique = pd.read_csv('./methyl_point_diff/pH5_diff_all.csv')
ph7_unique = pd.read_csv('./methyl_point_diff/pH7_diff_all.csv')

# annotated dataframes
annot = pd.read_csv("../hpy_annot.tsv",sep='\t')#hpy annotation file
annot = annot.drop(columns=["seqname", "source", "feature", "score", "frame", "Prot_Size(est)", "attribute"]) #clean
tss = pd.read_csv("../TSS.csv")
```

```python
ctrl_unique.head()
```

```python
counts = pd.DataFrame(columns=['gene', 'ctrl differences','arsS differences', 'pH 5 differences', 'pH 7 differences'])
counts['gene'] = annot['HP_number']
counts.head()
```

```python
for ind in counts.index: 
    counts.loc[ind, 'ctrl differences']=((ctrl_unique['start'].between(annot.loc[ind,'start'], annot.loc[ind,'end']+1))).sum()
    counts.loc[ind, 'arsS differences']=((ars_unique['start'].between(annot.loc[ind,'start'], annot.loc[ind,'end']+1))).sum()
    counts.loc[ind, 'pH 5 differences']=((ph5_unique['start'].between(annot.loc[ind,'start'], annot.loc[ind,'end']+1))).sum()
    counts.loc[ind, 'pH 7 differences']=((ph7_unique['start'].between(annot.loc[ind,'start'], annot.loc[ind,'end']+1))).sum()
```
